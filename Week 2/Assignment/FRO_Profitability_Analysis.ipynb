## Discount Cash Flows

Discount each location's annual net cash flow over its project life, subtract the initial investment, and calculate one NPV value for each location.

```python
npv_list = []

for _, location in df.iterrows():
    npv = -location["InitialInvestment"]

    for year in range(1, int(location["ProjectLife"]) + 1):
        discounted_cash_flow = (
            location["NetCashFlow"]
            / (1 + location["DiscountRate"]) ** year
        )
        npv += discounted_cash_flow

    npv_list.append(npv)

df["NPV"] = npv_list

df[["LocationName", "InitialInvestment", "NetCashFlow", "NPV"]]
```

## Break-Even Analysis

```python
df.columns.tolist()
```

```python
# Create a separate DataFrame for the break-even model
be_df = df.copy()

# Calculate break-even model results
be_df["AnnualRevenue"] = (
    be_df["AvgTransaction"]
    * be_df["ExpectedAnnualTransactions"]
)

be_df["VariableCosts"] = (
    be_df["AnnualRevenue"]
    * be_df["VariableCostPct"]
)

be_df["FixedCosts"] = (
    be_df["AnnualRent"]
    + be_df["AnnualLaborCost"]
    + be_df["OtherAnnualFixedCosts"]
)

be_df["ContributionMargin"] = (
    be_df["AnnualRevenue"]
    - be_df["VariableCosts"]
)

be_df["AnnualOperatingProfit"] = (
    be_df["ContributionMargin"]
    - be_df["FixedCosts"]
)

be_df["OperatingMargin"] = (
    be_df["AnnualOperatingProfit"]
    / be_df["AnnualRevenue"]
)

be_df["BreakEvenTransactions"] = (
    be_df["FixedCosts"]
    / (be_df["AvgTransaction"] * (1 - be_df["VariableCostPct"]))
)

be_df["BreakEvenRevenue"] = (
    be_df["FixedCosts"]
    / (1 - be_df["VariableCostPct"])
)

# Export the raw pandas results
be_df.to_excel("BE_Results.xlsx", index=False)

# Display the key results
be_df[[
    "LocationName",
    "AnnualRevenue",
    "VariableCosts",
    "FixedCosts",
    "ContributionMargin",
    "AnnualOperatingProfit",
    "OperatingMargin",
    "BreakEvenTransactions",
    "BreakEvenRevenue"
]]
```
